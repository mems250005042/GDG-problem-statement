```python
import os
import requests

pdf_path="apple_news_pdf.pdf"

url="https://raw.githubusercontent.com/mems250005042/Apple-News/main/Apple%20news%20pdf.pdf"

filename=pdf_path

response=requests.get(url)

if response.status_code==200:
  with open(filename, "wb") as file:
    file.write(response.content)
    print(f"file downloaded and saved as {filename}")

```

```python
import fitz
from tqdm.auto import tqdm

def text_formatter(text):
  cleaned_text = text.replace("\n"," ".strip())
  return cleaned_text

def open_and_read_pdf(pdf_path):
  doc=fitz.open(pdf_path)
  pages_and_texts=[]

  for page_number  , page in tqdm(enumerate(doc)):
    text=page.get_text()
    text=text_formatter(text)

    pages_and_texts.append({"page_number" : page_number , "page_char_count" : len(text) ,
                           "page_word_count" : len(text.split(" ")),
                           "page_sentence_count_raw" : len(text.split(". ")),
                           "text" : text})
  return pages_and_texts

pages_and_texts=open_and_read_pdf(pdf_path)
```

```python
from spacy.lang.en import English
nlp=English()
nlp.add_pipe("sentencizer")

for item in pages_and_texts:
  item["sentences"] = list(nlp(item["text"]).sents)
  item["sentences"]= [str(sentence) for sentence in item["sentences"]]
  item["page_sentence_count_spacy"]=len(item["sentences"])
```

```python
num_sentence_chunk=7

def split_list(input_list , slice_size):
  return [input_list[i : i+slice_size] for i in range(0,len(input_list), slice_size)]

for item in pages_and_texts:
  item["sentence_chunks"]=split_list(item["sentences"] , slice_size=num_sentence_chunk)
```

```python
pages_and_texts[5]["sentence_chunks"]
```

```python
import re
chunks=[]
for item in tqdm(pages_and_texts):
  for chunk in item["sentence_chunks"]:
    chunk_dict={}
    chunk_dict["page_number"]=item["page_number"]
    joined_sentence_chunk ="".join(chunk).replace("  "," ").strip()
    joined_sentence_chunk=re.sub(r'\.([A-Z])',r'. \1',joined_sentence_chunk)
    chunk_dict["chunk_text"]=joined_sentence_chunk

    chunk_dict["chunk_char_count"]=len(joined_sentence_chunk)
    chunk_dict["chunk_word_count"]=len(joined_sentence_chunk.split(" "))
    chunk_dict["chunk_token_count"]= len(joined_sentence_chunk)/4

    chunks.append(chunk_dict)
```

```python
def retrieve_relevant_resources(query ,embeddings=embeddings ,resources_to_return=10,embedding_model=embedding_model):
    query_embedding=embedding_model.encode(query  , convert_to_tensor=True)
    dot_scores=util.dot_score(query_embedding , embeddings)[0]
    scores , indices=torch.topk(dot_scores , k=resources_to_return)

    return scores , indices
```

```python

def print_top_results_and_scores(query , embeddings=embeddings, embedding_model=embedding_model,chunks_list=chunks_list ,resources_to_return =10):
    scores , indices =retrieve_relevant_resources(query)
    print(f'Query : {query}')
    print("results :")

    import numpy as np
    from numpy.linalg import norm

    query_embedding=embedding_model.encode(query)
    rerank_scores=[]
    for idx in indices:
      new_embedding=embeddings[idx]
      #cosine similarity

      new_score=np.dot(query_embedding , new_embedding)/(norm(query_embedding)*norm(new_embedding))
      rerank_scores.append((idx , new_score))

    rerank_scores=sorted(rerank_scores ,key=lambda x: x[1],reverse=True)
    for idx , score in rerank_scores:
      print(f"score :  {score}")
      print(chunks_list[idx])




```

```python
def build_prompt(query , context_items=context_items):
    """ augments query with text-based context from context_item"""
    base_prompt="""You are a financial analysis assistant.

Use ONLY the information explicitly provided in the context.
If a value is missing or not stated, say "Not provided in the data."
If the exact thing is missing , give info related to it .
focus on dates, don't give some other date's data for other dates .

if question specifices any date , you can include the data of that whole year
in which that date is present ,

Rules:
- Do not mix data from different years.
-At the end , add "Summary :" and provide summary of context which will include
your opinion on the given question.


Output rules:
- Write clear, grammatically correct sentences.
- Do not repeat the question.
- Do not include contradictions.
- If the data is insufficient, state that clearly and give answer related to
that question if not exact.
-don't just focus on keywords but also understand the question and what is
relates to it present in context.



Context:
{context}

User Question:
{query}

Instructions:
- Be concise and factual
- Mention dates when relevant
- Do not speculate
- Do not invent reasons
"""
    context="- "+"\n- ".join(i for i in context_items)
    base_prompt=base_prompt.format(context=context, query=query+"On the basis of the infotmation you have.")
    return base_prompt

```

```python
def ask(query):
    prompt=build_prompt(query)
    response=client.chat.completions.create(
    model='llama-3.1-8b-instant',
    messages=[{"role" : "user" , "content":prompt}],

    temperature=0.2)

    print(f"Query : \n {query}")
    print(f"Answer : ")
    output=response.choices[0].message.content
    output=output.replace("Based on the provided data","").replace(", " , "")
    print(output)
```
